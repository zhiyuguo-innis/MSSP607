### 3a. When does the code in a function execute?
### The code inside a function executes when the function is called, not when the function is defined. When a function is defined, Python stores the function. The statements inside the function run when we call the function.



```python
def hello():
    print("Hello!")

# Defining the function does not execute the print statement.
print("The function has been defined.")

# The function executes when it is called.
hello()
```

### 3b. How many global scopes are there in a Python program?
### How many local scopes？

### A Python program has one global scope.
### A local scope is created each time a function is called. Therefore, a program can have many local scopes during its execution, depending on how many functions are called

```python
x = "global"

def example():
    y = "local"
    print(x)
    print(y)

example()
```

### 3c. What happens to variables in a local scope when the function call returns?

### When a function call returns, the local scope is destroyed. The local variables normally cannot be accessed from outside the function.

```python
def example():
    local_variable = "Hello"
    print(local_variable)

example()

# This would cause an error because local_variable
# only exists inside the function.
# print(local_variable)
```

### 3d. How can you prevent a program from crashing when it gets an error?

We can use try and except statements to handle errors.
The code that might cause an error is placed inside the try block,
and the except block handles the error.

```python
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("That was not a valid number.")
```

```python
def collatz(number):
    # If the number is even, divide it by 2.
    if number % 2 == 0:
        return number // 2

    # If the number is odd, multiply by 3 and add 1.
    else:
        return 3 * number + 1
```

```python
# Ask the user to enter an integer.
print("Enter an integer:")

try:
    number = int(input())

   # Continue the sequence until the number becomes 1.
    while number != 1:
        number = collatz(number)
        print(number)

except ValueError:
    print("Please enter an integer.")
```
