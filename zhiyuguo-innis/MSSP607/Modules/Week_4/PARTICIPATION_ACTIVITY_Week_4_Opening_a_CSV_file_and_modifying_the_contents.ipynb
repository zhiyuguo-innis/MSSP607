````python
import pandas as pd

DATA_URL = (
    "https://raw.githubusercontent.com/zhiyuguo-innis/mssp607/main/"
    "zhiyuguo-innis/MSSP607/Modules/Week_4/parsed.csv"
)

df = pd.read_csv(DATA_URL)

print(df.shape)
df.head()
````

## a. 95th percentile of earthquake magnitude in Japan (magType `mb`)

````python
japan_mb = df[(df["parsed_place"] == "Japan") & (df["magType"] == "mb")]

print("Number of earthquakes:", len(japan_mb))
japan_mb["mag"].quantile(0.95)
````

**Answer:** Among the 50 `mb`-type earthquakes in Japan, the 95th percentile magnitude is **4.9**.

## b. Percentage of earthquakes in Indonesia coupled with tsunamis

````python
indonesia = df[df["parsed_place"] == "Indonesia"]

indonesia_total = len(indonesia)
indonesia_tsunamis = indonesia["tsunami"].sum()
percentage = indonesia_tsunamis / indonesia_total * 100

print("Total earthquakes:", indonesia_total)
print("Earthquakes with tsunamis:", indonesia_tsunamis)
print("Percentage:", round(percentage, 2), "%")
````

**Answer:** 34 of 147 earthquakes in Indonesia (**23.13%**) were coupled with tsunamis.

## c. Summary statistics for earthquakes in Nevada

````python
nevada = df[df["parsed_place"] == "Nevada"]

nevada.describe()
````

````python
nevada["mag"].describe()
````

## d. Ring of Fire indicator column
Notes on matching:
- **Exact matching** (`isin`) is used for most places, so `New Mexico` is **not** selected when matching `Mexico`.
- **Antarctica** does not appear as `"Antarctica"` in `parsed_place`. It shows up as ocean ridges (e.g. `Pacific-Antarctic Ridge`), so those are matched by the substring `"Antarctic"`.

````python
# Check how the tricky place names appear in the data
df[df["parsed_place"].str.contains("Antarctic|Mexico|Kermadec", na=False)]["parsed_place"].value_counts()
````

````python
ring_of_fire_places = [
    "Bolivia", "Chile", "Ecuador", "Peru", "Costa Rica", "Guatemala", "Mexico",
    "Japan", "Philippines", "Indonesia", "New Zealand", "Canada", "Fiji",
    "Alaska", "Washington", "California", "Russia", "Taiwan", "Tonga",
    "Kermadec Islands"
]

df["ring_of_fire"] = (
    df["parsed_place"].isin(ring_of_fire_places)
    | df["parsed_place"].str.contains("Antarctic", na=False)
)

# Sanity checks
print("New Mexico flagged?", df.loc[df["parsed_place"] == "New Mexico", "ring_of_fire"].any())
print("Antarctic rows flagged:", df.loc[df["parsed_place"].str.contains("Antarctic", na=False), "ring_of_fire"].sum())

df[["parsed_place", "ring_of_fire"]].head(10)
````

## e. Number of earthquakes inside vs. outside the Ring of Fire

````python
df["ring_of_fire"].value_counts()
````

**Answer:** **7,188** earthquakes occurred in Ring of Fire locations and **2,144** occurred outside them.

## f. Tsunami count along the Ring of Fire

````python
ring_of_fire_tsunamis = df[df["ring_of_fire"] & (df["tsunami"] == 1)]

len(ring_of_fire_tsunamis)
````

**Answer:** **45** earthquakes along the Ring of Fire were associated with tsunamis.
