```python
import pandas as pd

df = pd.read_csv("parsed.csv")

df.head()
```

quetsion a：The 95th percentile of earthquake magnitude in Japan for earthquakes with an mb magnitude type is 4.9.

```python
japan_mb = df[
    (df["parsed_place"] == "Japan") &
    (df["magType"] == "mb")
]

japan_mb["mag"].quantile(0.95)
```

```python
len(japan_mb)
```

question b: 23.13% of the earthquakes in Indonesia were coupled with tsunamis.

```python
indonesia = df[df["parsed_place"] == "Indonesia"]

indonesia["tsunami"].mean() * 100
```

```python
round(indonesia["tsunami"].mean() * 100, 2)
```

```python
indonesia_tsunamis = indonesia["tsunami"].sum()
indonesia_total = len(indonesia)

percentage = indonesia_tsunamis / indonesia_total * 100

print("Total earthquakes:", indonesia_total)
print("Earthquakes with tsunamis:", indonesia_tsunamis)
print("Percentage:", round(percentage, 2), "%")
```
